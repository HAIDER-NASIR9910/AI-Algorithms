```python
import math
import time
from collections import deque
import heapq
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd   # for dataframe

# ===================== GRID AND COSTS =====================
grid = [
    ['B','P','P','P','F','F','F','X','X','P','P','P','F','F','H','P','F','P','P','P'],
    ['P','P','F','F','H','F','F','X','X','P','F','H','H','P','P','F','P','F','F','P'],
    ['P','F','F','H','F','F','P','X','X','P','H','H','F','F','H','P','F','P','P','P'],
    ['P','F','H','H','F','P','P','X','X','X','P','F','H','H','P','P','P','F','P','P'],
    ['F','H','H','F','F','P','R','X','X','X','P','F','F','H','H','P','F','P','P','F'],
    ['H','H','F','F','P','P','P','P','P','P','P','F','H','F','F','H','P','P','P','P'],
    ['P','P','F','H','H','F','F','F','P','V','P','P','P','H','F','P','P','F','P','P'],
    ['F','F','H','F','F','F','H','H','F','P','P','H','P','F','P','P','F','P','H','P'],
    ['P','F','H','H','F','P','P','P','P','P','P','P','F','H','F','F','P','P','P','P'],
    ['P','P','F','F','P','R','H','P','P','P','P','P','P','H','H','F','F','P','F','P'],
    ['P','H','H','F','F','F','F','F','P','P','P','H','F','F','P','P','H','P','P','P'],
    ['P','F','F','P','P','P','H','H','F','F','P','P','P','F','F','P','F','F','H','P'],
    ['P','P','H','F','F','P','P','P','H','H','P','P','P','P','F','H','P','P','F','P'],
    ['P','F','H','H','P','P','F','H','F','P','P','H','P','F','F','H','P','V','P','P'],
    ['P','P','F','P','P','P','H','F','H','P','F','P','P','P','H','F','P','P','P','P'],
    ['P','P','P','P','F','P','F','P','H','P','P','P','P','F','P','H','P','F','F','P'],
    ['P','F','H','P','P','P','F','F','P','P','P','P','F','H','P','P','V','P','P','P'],
    ['P','F','F','H','P','P','H','F','F','P','H','P','P','P','P','P','F','H','P','P'],
    ['P','P','P','P','P','H','P','P','P','H','F','H','P','V','P','R','P','P','P','P'],
    ['P','P','P','F','P','P','P','F','F','P','P','P','H','F','H','P','P','P','V','P']
]

costs = {'P':1, 'F':2, 'H':3, 'B':0, 'V':1, 'R':1, 'X':math.inf}
fuel_limit = 25

rows, cols = len(grid), len(grid[0])

# ===================== HELPER FUNCTIONS =====================
def get_neighbors(r, c):
    moves = [(1,0), (-1,0), (0,1), (0,-1)]
    for dr, dc in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] != 'X':
            yield nr, nc

def terrain_cost(r, c):
    return costs.get(grid[r][c], math.inf)

# ===================== SEARCH ALGORITHMS =====================
def bfs(start, goal):
    q = deque([(start, [start])])
    visited = set([start])
    while q:
        (r, c), path = q.popleft()
        if (r, c) == goal:
            return path, visited
        for nr, nc in get_neighbors(r, c):
            if (nr, nc) not in visited:
                visited.add((nr, nc))
                q.append(((nr, nc), path + [(nr, nc)]))
    return None, visited

def dfs(start, goal):
    stack = [(start, [start])]
    visited = set([start])
    while stack:
        (r, c), path = stack.pop()
        if (r, c) == goal:
            return path, visited
        for nr, nc in get_neighbors(r, c):
            if (nr, nc) not in visited:
                visited.add((nr, nc))
                stack.append(((nr, nc), path + [(nr, nc)]))
    return None, visited

def heuristic(a, b, method):
    (x1, y1), (x2, y2) = a, b
    if method == "euclidean":
        return math.sqrt((x1-x2)**2 + (y1-y2)**2)
    elif method == "custom":
        return 1.5 * (abs(x1-x2) + abs(y1-y2)) + 2
    else:
        return abs(x1-x2) + abs(y1-y2)

def astar(start, goal, method="manhattan"):
    pq = [(0, start, [start])]
    g = {start: 0}
    visited = set()

    while pq:
        f, (r, c), path = heapq.heappop(pq)
        if (r, c) == goal:
            return path, visited
        if (r, c) in visited:
            continue
        visited.add((r, c))

        for nr, nc in get_neighbors(r, c):
            new_g = g[(r, c)] + terrain_cost(nr, nc)
            if new_g <= fuel_limit and ((nr, nc) not in g or new_g < g[(nr, nc)]):
                g[(nr, nc)] = new_g
                f = new_g + heuristic((nr, nc), goal, method)
                heapq.heappush(pq, (f, (nr, nc), path + [(nr, nc)]))
    return None, visited

# ===================== RUN TEST =====================
base = (0, 0)
villages = [(6,9), (13,17), (16,16), (18,13), (19,18)]

algorithms = {
    "BFS": bfs,
    "DFS": dfs,
    "A*_Manhattan": lambda s, g: astar(s, g, "manhattan"),
    "A*_Euclidean": lambda s, g: astar(s, g, "euclidean"),
    "A*_Custom": lambda s, g: astar(s, g, "custom")
}

print("\nDrone Pathfinding Comparison\n")

# store results
results = []

for name, algo in algorithms.items():
    start_time = time.time()
    time.sleep(0.5)
    path, visited = algo(base, villages[0])
    end_time = time.time()

    if path:
        total_cost = sum(terrain_cost(r, c) for r, c in path)
        steps = len(path)
        duration = round(end_time - start_time, 4)
        print(name + ": Found path to V1, Cost=" + str(total_cost) +
              ", Steps=" + str(steps) +
              ", Time=" + str(duration) + "s")

        # store data
        results.append([name, total_cost, steps, duration])

        # heatmap
        heatmap = np.zeros((rows, cols))
        for r, c in visited:
            heatmap[r][c] = 0.5
        for r, c in path:
            heatmap[r][c] = 1.0

        plt.figure(figsize=(6,6))
        plt.title("Heatmap - " + name)
        plt.imshow(heatmap, cmap='YlOrRd')
        plt.colorbar(label='Exploration Level')
        plt.show()
    else:
        print(name + ": No path found!")

print("\nEnd of Simulation")

# ===================== DATAFRAME RESULTS =====================
df = pd.DataFrame(results, columns=["Algorithm", "Cost", "Steps", "Time (s)"])
df = df.sort_values(by="Cost")
print("\nAlgorithm Comparison Table:")
print(df)
```

This Python code compares **five pathfinding algorithms** — BFS, DFS, A* (with Manhattan, Euclidean, and Custom heuristics) — to find the most efficient route for a **drone navigating a 20×20 terrain grid**. Each grid cell has a **terrain type** (Plain, Forest, Hill, etc.) with different movement costs. The program defines helper functions to get valid neighbors and calculate terrain cost. Each algorithm searches for a path from the **base (B)** to a **village (V1)** while staying within a **fuel limit of 25**. During execution, it measures **path cost, number of steps, and execution time** for every algorithm. It also visualizes each search as a **heatmap**, showing explored and final path cells. Finally, it compiles the results into a **DataFrame table**, sorted by path cost, for an easy performance comparison.
