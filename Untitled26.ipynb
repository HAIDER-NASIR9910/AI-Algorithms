```python
# --- Three-player Max-N algorithm (AIMA Figure 5.4) ---

# Game tree structure
game_tree = {
    'A': ['B_left', 'B_right'],
    'B_left': ['C_left', 'C_right'],
    'C_left': ['t1', 't2'],
    'C_right': ['t3', 't4'],
    'B_right': ['C3', 'C4'],
    'C3': ['t5', 't6'],
    'C4': ['t7', 't8'],
}

# Leaf utilities
utilities = {
    't1': (1,2,6), 't2': (4,2,3),
    't3': (6,1,2), 't4': (7,4,1),
    't5': (5,1,1), 't6': (1,5,2),
    't7': (7,7,1), 't8': (5,4,5),
}

players = ['A', 'B', 'C']

def maxn(node, depth=0):
    """Return utility vector from this node using Max-N."""
    # Terminal node
    if node in utilities:
        return utilities[node]

    player = players[depth % 3]
    best_val = None
    best_child = None

    for child in game_tree[node]:
        val = maxn(child, depth + 1)

        # initialize
        if best_val is None:
            best_val, best_child = val, child
        else:
            if val[players.index(player)] > best_val[players.index(player)]:
                best_val, best_child = val, child
            # Equal? then keep the first one (tie-break → leftmost)
            # no update on equality

    print(f"{node:<7} | Player {player} chooses {best_child} → {best_val}")
    return best_val

result = maxn('A')
print("\nFinal result at root:", result)
```
