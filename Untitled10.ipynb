```python
from collections import deque
def bfs(graph,start,goal):
  visited=set()
  qeue=deque([[start]])
  while qeue:
    path=qeue.popleft()
    node=path[-1]
    if node==goal:
      return path
    if node not in visited:
      visited.add(node)
      for neighbours in graph.get(node,[]):
        new_path=path+[neighbours]
        qeue.append(new_path)
  return None

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print(bfs(graph, 'A', 'F'))  # Output: ['A', 'C', 'F']
```

```python
import heapq
def uniform_cost_search(graph,start,goal):
  visited=set()
  pq=[(0,[start])]
  while pq:
    cost,path=heapq.heappop(pq)
    node=path[-1]
    if node==goal:
      return cost,path
    if node not in visited:
      visited.add(node)
      for neighbour,cost_path in graph.get(node,[]):
        new_cost=cost+cost_path
        new_path=path+[neighbour]
        heapq.heappush(pq,(new_cost,new_path))

  return None

graph = {
    'A': [('B', 2), ('C', 5)],
    'B': [('D', 4), ('E', 1)],
    'C': [('F', 2)],
    'D': [],
    'E': [('F', 3)],
    'F': []
}

print(uniform_cost_search(graph, 'A', 'F'))
```

```python
def dls(graph,node,goal,limit):
  visited=set()
  if node==goal:
    return [node]
  if limit==0:
    return None
  for neighbour in graph.get(node,[]):
      path=dls(graph,neighbour,goal,limit-1)
      if path:
        return [node]+path
  return None

def iddfs(graph,start,goal,max_depth):
  for depth in range(max_depth+1):
    path=dls(graph,start,goal,depth)
    if path:
      return path

  return None




graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print(iddfs(graph, 'A', 'F', 1))  # None (too shallow)
print(iddfs(graph, 'A', 'F', 2))  # ['A', 'C', 'F']
print(iddfs(graph, 'A', 'F', 3))  # ['A', 'B', 'E', 'F'] (depending on order)
```
