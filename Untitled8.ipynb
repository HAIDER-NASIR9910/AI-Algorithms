```python
from collections import deque

def bidirectional_search(graph, start, goal):
    if start == goal:
        return [start]

    # Queues for forward and backward search
    queue_f = deque([[start]])
    queue_b = deque([[goal]])

    # Visited dictionaries store paths
    visited_f = {start: [start]}
    visited_b = {goal: [goal]}

    while queue_f and queue_b:
        # --- Forward expansion ---
        path_f = queue_f.popleft()
        node_f = path_f[-1]
        for neighbour in graph.get(node_f, []):
            if neighbour not in visited_f:
                new_path = path_f + [neighbour]
                visited_f[neighbour] = new_path
                queue_f.append(new_path)
            if neighbour in visited_b:  # Meeting point
                return visited_f[neighbour] + visited_b[neighbour][-2::-1]

        # --- Backward expansion ---
        path_b = queue_b.popleft()
        node_b = path_b[-1]
        for neighbour in graph.get(node_b, []):
            if neighbour not in visited_b:
                new_path = path_b + [neighbour]
                visited_b[neighbour] = new_path
                queue_b.append(new_path)
            if neighbour in visited_f:  # Meeting point
                return visited_f[neighbour] + visited_b[neighbour][-2::-1]

    return None

# Example undirected graph
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# Test
print(bidirectional_search(graph, 'A', 'F'))
```
