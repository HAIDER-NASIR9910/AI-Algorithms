```python
# Simple Beam Search Algorithm

# Example graph
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['E', 'F'],
    'C': ['G'],
    'D': ['H'],
    'E': [],
    'F': [],
    'G': [],
    'H': []
}

# Heuristic values (higher = better)
heuristic = {
    'A': 1,
    'B': 5,
    'C': 3,
    'D': 4,
    'E': 8,
    'F': 6,
    'G': 7,
    'H': 5
}

def beam_search(start, beam_width):
    # Start with the first node
    current_nodes = [start]

    print("Start from:", start)

    # Continue until no nodes left
    while current_nodes:
        print("Current level:", current_nodes)

        # Generate all possible next nodes (children)
        next_nodes = []
        for node in current_nodes:
            next_nodes.extend(graph[node])

        # If no next nodes, stop
        if not next_nodes:
            print("No more nodes to explore.")
            break

        # Sort by heuristic value (keep only best few)
        next_nodes.sort(key=lambda n: heuristic[n], reverse=True)
        current_nodes = next_nodes[:beam_width]

        print("Selected best nodes:", current_nodes)

    print("Search finished.")

# Run the algorithm
beam_search('A', 2)   # beam_width = 2 (keep best 2 at each level)
```

```python
# Simple A* Search Algorithm

graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 2, 'C': 2},
    'C': {'G': 3},
    'D': {'G': 5},
    'G': {}
}

# Heuristic values (estimated cost to reach goal G)
heuristic = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 1,
    'G': 0
}

def a_star(start, goal):
    open_list = [start]     # nodes to explore
    closed_list = []        # nodes already visited
    g_cost = {start: 0}     # actual cost from start to node
    parent = {start: None}  # to keep track of path

    while open_list:
        # Find node with lowest f = g + h
        current = min(open_list, key=lambda x: g_cost[x] + heuristic[x])

        print("Visiting:", current)

        # If goal found, stop
        if current == goal:
            break

        open_list.remove(current)
        closed_list.append(current)

        # Check all neighbors
        for neighbor, cost in graph[current].items():
            if neighbor in closed_list:
                continue

            new_g = g_cost[current] + cost

            # If neighbor not in open list or found a cheaper path
            if neighbor not in open_list or new_g < g_cost.get(neighbor, 999):
                g_cost[neighbor] = new_g
                parent[neighbor] = current
                if neighbor not in open_list:
                    open_list.append(neighbor)

    # Reconstruct path
    path = []
    node = goal
    while node is not None:
        path.insert(0, node)
        node = parent.get(node)

    print("Shortest Path:", path)
    print("Total Cost:", g_cost[goal])

# Run the algorithm
a_star('A', 'G')
```

```python
import math
import random

# Function we want to minimize (smaller is better)
def function(x):
    return x**2 + 4 * math.sin(5 * x)

def simulated_annealing():
    # Step 1: Start with a random solution
    current_x = random.uniform(-10, 10)
    current_value = function(current_x)

    # Step 2: Set initial temperature and cooling settings
    temperature = 100
    cooling_rate = 0.95
    min_temp = 0.1

    print("Start at x =", round(current_x, 2), "value =", round(current_value, 2))

    # Step 3: Repeat until cooled down
    while temperature > min_temp:
        # Step 4: Make a small random move
        new_x = current_x + random.uniform(-1, 1)
        new_value = function(new_x)

        # Step 5: Calculate the change
        diff = new_value - current_value

        # Step 6: If new is better, take it
        if diff < 0:
            current_x = new_x
            current_value = new_value
        else:
            # If worse, sometimes still accept it
            chance = math.exp(-diff / temperature)
            if random.random() < chance:
                current_x = new_x
                current_value = new_value

        # Step 7: Cool down the temperature
        temperature = temperature * cooling_rate

        print("x =", round(current_x, 2), "value =", round(current_value, 2), "temp =", round(temperature, 2))

    print("\nBest found x =", round(current_x, 2), "with value =", round(current_value, 2))

# Run the algorithm
simulated_annealing()
```
