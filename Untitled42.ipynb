```python
!pip install yt-dlp

import cv2
from ultralytics import YOLO
import yt_dlp
import os

# --- User input: Replace with your YouTube video URL ---
youtube_url = 'https://www.youtube.com/watch?v=dQw4w9WgXcQ' # Placeholder: Please replace with your actual YouTube link
output_filename = 'detected_video.mp4'

# 1. Download the YouTube video
print(f"Downloading video from {youtube_url}...")

# yt_dlp options to download the best quality video as an mp4 file
ydl_opts = {
    'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/best[ext=mp4]',
    'outtmpl': 'input_video.mp4',
    'noplaylist': True,
    'quiet': True,
}

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    info_dict = ydl.extract_info(youtube_url, download=True)
    video_path = ydl.prepare_filename(info_dict)

print(f"Video downloaded to: {video_path}")

# 2. Load the pre-trained YOLO model (already loaded in kKtOtU4LFL70, reusing it)
# model = YOLO("yolo11m.pt") # This line is commented out as the model is already loaded in cell kKtOtU4LFL70

# Check if the model is loaded, if not, load it. Assuming it's in the global scope from cell kKtOtU4LFL70
try:
    if 'model' not in globals() or not isinstance(model, YOLO):
        model = YOLO("yolo11m.pt")
except NameError:
    model = YOLO("yolo11m.pt")

# 3. Apply human detection and save the video
print("Applying human detection...")

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print(f"Error: Could not open video file {video_path}")
else:
    # Get video properties
    frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = int(cap.get(cv2.CAP_PROP_FPS))

    # Define the codec and create VideoWriter object
    # Using 'mp4v' for .mp4 files, or 'XVID' for .avi
    fourcc = cv2.VideoWriter_fourcc(*'mp4v') # Codec for .mp4 output
    out = cv2.VideoWriter(output_filename, fourcc, fps, (frame_width, frame_height))

    if not out.isOpened():
        print(f"Error: Could not create output video writer for {output_filename}")
    else:
        while True:
            ret, frame = cap.read()
            if not ret:
                break

            # Perform inference
            results = model(frame, stream=True)

            # Iterate over results to draw bounding boxes for 'person' class
            for r in results:
                # Get detection boxes
                boxes = r.boxes
                for box in boxes:
                    # Get class name
                    cls = int(box.cls[0])
                    class_name = model.names[cls]

                    if class_name == 'person':
                        # Get bounding box coordinates
                        x1, y1, x2, y2 = map(int, box.xyxy[0])
                        # Draw rectangle
                        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2) # Green box
                        # Put label
                        confidence = float(box.conf[0])
                        label = f"Person: {confidence:.2f}"
                        cv2.putText(frame, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

            # Write the processed frame to the output video
            out.write(frame)

        cap.release()
        out.release()
        print(f"Processed video saved as {output_filename}")
        print(f"You can now download '{output_filename}' from your Colab files sidebar.")

    # Clean up the downloaded input video
    if os.path.exists(video_path):
        os.remove(video_path)
        print(f"Removed temporary input video: {video_path}")
```
