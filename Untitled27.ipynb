```python
#HAIDER NASIR
#B24S0308AI088
#AI-BLUE
import random
import time
# TASK 1:
floors = [1, 3, 4, 6, 8, 10, 14]
start_floor = 3

def total_distance(f):
    return sum(abs(f - x) for x in floors)

print("TASK 1: Elevator Optimization ")
print("Floors where people are:", floors)
print("Start from floor:", start_floor)

# Simple Hill Climbing
current = start_floor
start_time = time.time()
while True:
    left = total_distance(current - 1)
    current_distance = total_distance(current)
    right = total_distance(current + 1)

    if left < current_distance:
        current -= 1
    elif right < current_distance:
        current += 1
    else:
        break
end_time = time.time()
print("\nSimple Hill Climbing:")
print("Optimal floor:", current, "Total distance:", total_distance(current))
print("Execution time:", round(end_time - start_time, 6), "seconds")

# Steepest-Ascent Hill Climbing
current = start_floor
start_time = time.time()
while True:
    neighbors = [current - 1, current + 1]
    distances = [total_distance(f) for f in neighbors]
    min_distance = min(distances)
    if min_distance < total_distance(current):
        current = neighbors[distances.index(min_distance)]
    else:
        break
end_time = time.time()
print("Steepest-Ascent Hill Climbing:")
print("Optimal floor:", current, "Total distance:", total_distance(current))
print("Execution time:", round(end_time - start_time, 6), "seconds")

#Stochastic Hill Climbing
current = start_floor
start_time = time.time()
for i in range(50):
    random_floor = random.choice([f for f in floors if f != current])
    if total_distance(random_floor) < total_distance(current):
        current = random_floor
end_time = time.time()
print("\nStochastic Hill Climbing:")
print("Final floor:", current, "Total distance:", total_distance(current))
print("Execution time:", round(end_time - start_time, 6), "seconds")
```

Task 1: Elevator Optimal Stop Floor

Each floor in the building is treated as a possible state for the elevator.

The cost function calculates the total walking distance of all people from the elevator’s current floor.

Simple Hill Climbing moves the elevator one floor at a time toward decreasing distance until no improvement.

Steepest-Ascent Hill Climbing checks all neighboring floors and chooses the one with the lowest total distance.

Stochastic Hill Climbing randomly selects floors and moves only if it reduces the total distance; execution time and optimal floor are compared.
