```python
from collections import deque
from typing import Deque
def bfs(graph,start,goal):
  visited=set()
  queue=deque([[start]])
  while queue:
    path=queue.popleft()
    node=path[-1]
    if node==goal:
      return path

    if node not in visited:
      visited.add(node)
      for neighbours in graph.get(node,[]):
        new_path=path+[neighbours]
        queue.append(new_path)
  return None
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print(bfs(graph, 'A', 'F'))  # Output: ['A', 'C', 'F']


```

```python
import heapq
def uniform_cost_search(graph,start,goal):
  visited=set()
  pq=[(0,[start])]
  while pq:
    cost,path=heapq.heappop(pq)
    node=path[-1]
    if node==goal:
      return cost,path
    if node not in visited:
      visited.add(node)
      for neighbour,cost_edge in graph.get(node,[]):
        new_cost=cost+cost_edge
        new_path=path+[neighbour]
        heapq.heappush(pq,(new_cost,new_path))

  return None

graph = {
    'A': [('B', 2), ('C', 5)],
    'B': [('D', 4), ('E', 1)],
    'C': [('F', 2)],
    'D': [],
    'E': [('F', 3)],
    'F': []
}

print(uniform_cost_search(graph, 'A', 'F'))
```

```python
from collections import deque

def bidirectional_search(graph, start, goal):
    if start == goal:
        return [start]

    # Forward and backward queues
    queue_start = deque([[start]])
    queue_goal = deque([[goal]])

    # Visited sets
    visited_start = {start: [start]}
    visited_goal = {goal: [goal]}

    while queue_start and queue_goal:
        # Expand from start side
        path_s = queue_start.popleft()
        node_s = path_s[-1]

        if node_s in visited_goal:  # Meeting point
            return path_s + visited_goal[node_s][::-1][1:]

        for nbr in graph.get(node_s, []):
            if nbr not in visited_start:
                visited_start[nbr] = path_s + [nbr]
                queue_start.append(path_s + [nbr])

        # Expand from goal side (reverse search)
        path_g = queue_goal.popleft()
        node_g = path_g[-1]

        if node_g in visited_start:  # Meeting point
            return visited_start[node_g] + path_g[::-1][1:]

        for node, nbrs in graph.items():   # scan reverse edges
            if node_g in nbrs and node not in visited_goal:
                visited_goal[node] = path_g + [node]
                queue_goal.append(path_g + [node])

    return None


# Example graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print(bidirectional_search(graph, 'A', 'F'))
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Define the graph
graph = {
    1: [2, 3],
    2: [3, 5],
    3: [4],
    4: [6],
    5: [],
    6: []
}

# Create a directed graph
G = nx.DiGraph()

# Add edges
for node, neighbors in graph.items():
    for neighbor in neighbors:
        G.add_edge(node, neighbor)

# Draw the graph
plt.figure(figsize=(6, 5))
nx.draw(G, with_labels=True, node_color="skyblue", node_size=1500,
        arrowsize=20, font_size=12, font_weight="bold", edge_color="gray")

plt.title("Graph Visualization (Directed)", fontsize=14)
plt.show()
```
