```python
print("Enter the amount invested in each ETF:")
spy = float(input("SPY"))
qqq = float(input("QQQ"))
eem = float(input("EEM"))
vxx = float(input("VXX"))
total_amount = spy + qqq + eem + vxx
spy_percentage = (spy / total_amount) * 100
qqq_percentage = (qqq / total_amount) * 100
eem_percentage = (eem / total_amount) * 100
vxx_percentage = (vxx / total_amount) * 100
print("PERCENTAGE")
print("SPY",spy_percentage)
print("QQQ",qqq_percentage)
print("EEM",eem_percentage)
print("VXX",vxx_percentage)
print("TOTAL AMOUNT INVESTED",total_amount)
```

```python
cycling_calories = 200
running_calories = 475
swimming_calories = 275
calories_per_kg = 3500
calories_per_pound = calories_per_kg / 2.20462
print("Enter the number of hours spent at each activity:")
cycling_hours = float(input("Cycling"))
running_hours = float(input("Running"))
swimming_hours = float(input("Swimming"))
total_calories = (cycling_calories * cycling_hours) + (running_calories * running_hours) + (swimming_calories * swimming_hours)

pounds_worked_off = total_calories / calories_per_pound
print("Total calories burned",total_calories)
print("Pounds worked off",pounds_worked_off)
```

```python
age = int(input("Enter your age: "))
resting_heart_rate = int(input("Enter your resting heart rate beats per minute"))
training_heart_rate = 0.7 * (220 - age) + 0.3 * resting_heart_rate
print(f"Your training heart rate is",training_heart_rate, "beats per minute")
```
