```python
from collections import deque
def bfs(graph,start,goal):
  visited=set()
  qeue=deque([[start]])
  while qeue:
    path=qeue.popleft()
    node=path[-1]
    if node==goal:
      return path
    if node not in visited:
      visited.add(node)
      for neighbour in graph.get(node,[]):
        new_path=path+[neighbour]
        qeue.append(new_path)
  return None

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print(bfs(graph, 'A', 'F'))  # Output: ['A', 'C', 'F']
```

```python
def dls(graph,node,goal,limit):
  visited=set()
  if limit==0:
    return
  if node==goal:
    return [node]

  if node not in visited:
    visited.add(node)
    for neighbour in graph.get(node,[]):
      path=dls(graph,neighbour,goal,limit-1)
      if path:
        return [node]+path
  return None



graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print(dls(graph, 'A', 'F',3))  # Output: ['A', 'C', 'F']
```

```python
import heapq
def ucs(graph,start,goal):
  visited=set()
  qeue=[(0,[start])]
  while qeue:
    cost,path=heapq.heappop(qeue)
    node=path[-1]
    if node==goal:
      return cost,path
    if node not in visited:
      visited.add(node)
      for neighbour,edge_cost in graph.get(node,[]):
        new_path=path+[neighbour]
        new_cost=cost+edge_cost
        heapq.heappush(qeue,(new_cost,new_path))
  return None
graph = {
    'A': [('B', 2), ('C', 5)],
    'B': [('D', 4), ('E', 1)],
    'C': [('F', 2)],
    'D': [],
    'E': [('F', 3)],
    'F': []
}

print(ucs(graph, 'A', 'F'))
```

```python
def dls(graph,node,goal,limit):
  visited=set()
  if node==goal:
    return [node]
  if limit==0:
    return
  if node not in visited:
    visited.add(node)
    for neighbour in graph.get(node,[]):
      path=dls(graph,neighbour,goal,limit-1)
      if path:
        return [node]+path
  return None
def iddfs(graph,start,goal,deapth):
  for max_depth in range(deapth):
    path=dls(graph,start,goal,max_depth+1)
    if path:
      return path
  return None


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

print(iddfs(graph, 'A', 'F',3))  # Output: ['A', 'C', 'F']
```

```python
#Haider Nasir
#B24S0308AI088
#AI-BLUE
#LAB#-8
from queue import PriorityQueue

def a_star_search(flights, heuristic, start, goal):
    open_list = PriorityQueue()
    open_list.put((0 + heuristic[start], 0, [start]))  # (f(n), g(n), path)
    visited = set()

    while not open_list.empty():
        f, g, path = open_list.get()
        current = path[-1]

        # Goal reached
        if current == goal:
            return path, g

        if current in visited:
            continue
        visited.add(current)

        for neighbor, cost in flights[current].items():
            new_g = g + cost
            f_new = new_g + heuristic[neighbor]
            open_list.put((f_new, new_g, path + [neighbor]))

    return None, float('inf')

# Example data
flights = {
    'A': {'B': 100, 'C': 300},
    'B': {'C': 100, 'D': 200},
    'C': {'D': 100, 'E': 400},
    'D': {'E': 100},
    'E': {}
}

heuristic = {'A': 400, 'B': 300, 'C': 150, 'D': 100, 'E': 0}

start, goal = 'A', 'E'

path, cost = a_star_search(flights, heuristic, start, goal)

print("Cheapest route:", path)
print("Total cost:", cost)
```
