```python
def dls(graph,node,goal,limit):
    if node==goal:
        return [node]
    if limit==0:
        return
    for neighbours in graph.get(node,[]):
        path=dls(graph,neighbours,goal,limit-1)  #its doing recursive call here
        if path:                                 #if the path found it will return
            return [node]+path

    return None
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G':[]
}
print(dls(graph,'A','G',3))
```

```python
from collections import deque

def reverse_graph(graph):
    rev_graph = {node: [] for node in graph}
    for node, neighbours in graph.items():
        for n in neighbours:
            rev_graph[n].append(node)
    return rev_graph


def bidirectional_search(graph, start, goal):
    if start == goal:
        return [start]

    # Create reverse graph for backward search
    rev_graph = reverse_graph(graph)

    queue_s = deque([[start]])  # forward search
    queue_g = deque([[goal]])   # backward search

    visited_s = {start}
    visited_g = {goal}

    while queue_s and queue_g:
        # --- Forward Search ---
        path_s = queue_s.popleft()
        node_s = path_s[-1]

        for neighbour in graph.get(node_s, []):
            if neighbour not in visited_s:
                new_path_s = path_s + [neighbour]
                queue_s.append(new_path_s)
                visited_s.add(neighbour)

                # Check intersection
                if neighbour in visited_g:
                    meeting_point = neighbour
                    print(f"Meeting point found at: {meeting_point}")
                    # Get backward path that starts from meeting point
                    for path in queue_g:
                        if path[0] == meeting_point:
                            full_path = new_path_s[:-1] + path[::-1]
                            print("Path from A to G:", full_path)
                            return full_path

        # --- Backward Search ---
        path_g = queue_g.popleft()
        node_g = path_g[-1]

        for neighbour in rev_graph.get(node_g, []):
            if neighbour not in visited_g:
                new_path_g = [neighbour] + path_g
                queue_g.append(new_path_g)
                visited_g.add(neighbour)

                if neighbour in visited_s:
                    meeting_point = neighbour
                    print(f"Meeting point found at: {meeting_point}")
                    for path in queue_s:
                        if path[-1] == meeting_point:
                            full_path = path + new_path_g[1:]
                            print("Path from A to G:", full_path)
                            return full_path
    print("No meeting point found.")
    return None


# City intersections (Graph)
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': ['G'],
    'G': []
}

bidirectional_search(graph, 'A', 'G')
```
