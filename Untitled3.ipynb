```python
class Book:
    def __init__(self, title, author, pages, available=True):
        self.title = title
        self.author = author
        self.pages = pages
        self.available = available

    def __str__(self):
        return f"Book name: {self.title}, Author: {self.author}, Pages: {self.pages}, Available: {self.available}"

    def __len__(self):
        return self.pages

    def print(self):
        print(self.__str__())


from abc import ABC, abstractmethod

class Person(ABC):
    @abstractmethod
    def get_role(self):
        pass


class Member(Person):
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self.__borrowed_books = []

    def borrow_book(self, book: Book):
        if book.available:
            self.__borrowed_books.append(book)
            book.available = False
            print(f"{book.title} has been borrowed.")
        else:
            print(f"{book.title} is not available.")

    def return_book(self, book: Book):
        if book in self.__borrowed_books:
            self.__borrowed_books.remove(book)
            book.available = True
            print(f"{book.title} has been returned.")
        else:
            print("This book was not borrowed by you.")

    def list_borrowed_books(self):
        if self.__borrowed_books:
            print("Borrowed books:")
            for book in self.__borrowed_books:
                print(f"- {book.title}")
        else:
            print("No books borrowed.")

    def get_role(self):
        return "Member"

    def print(self):
        print(f"Name: {self.name}, Member ID: {self.member_id}")
        self.list_borrowed_books()


class Librarian(Member):
    def __init__(self, name, member_id):
        super().__init__(name, member_id)

    def add_book(self, book_list, book: Book):
        book_list.append(book)
        print(f"{book.title} added to the library.")

    def remove_book(self, book_list, book: Book):
        if book in book_list:
            book_list.remove(book)
            print(f"{book.title} removed from the library.")
        else:
            print("Book not found in library.")

    def get_role(self):
        return "Librarian"
book1 = Book("Harry Potter", "J.K. Rowling", 350, True)
book2 = Book("Percy Jackson", "Rick Riordan", 400, True)

member1 = Member("Haider", 123)
librarian1 = Librarian("Admin", 1)

library_books = [book1, book2]

print(member1.get_role())
print(librarian1.get_role())

member1.borrow_book(book1)
member1.list_borrowed_books()

librarian1.remove_book(library_books, book2)
librarian1.add_book(library_books, Book("New Book", "Author X", 200, True))


```

This program manages a library with books, members, and librarians.

A Book class stores details and uses magic methods: __str__ for readable output and __len__ for page count.

An abstract Person class forces subclasses to define get_role(), showing abstraction.

A Member class (inherits Person) can borrow, return, and list books. Borrowed books are kept private, showing encapsulation.

A Librarian class (inherits Member) adds or removes books, showing inheritance with extra functionality.

Polymorphism is used because both Member and Librarian define get_role() but return different results.

```python
class Vehicle:
    def __init__(self, make, model, year, rent_price):
        self.make = make
        self.model = model
        self.year = year
        self.rent_price = rent_price

    def calculate_rent(self, days):
        return self.rent_price * days


class Car(Vehicle):
    pass


class Bike(Vehicle):
    pass
make = input("Enter the make of the vehicle: ")
model = input("Enter the model of the vehicle: ")
year = int(input("Enter the year of the vehicle: "))
rent_price = float(input("Enter the rent price of the vehicle per day: "))

vehicle1 = Vehicle(make, model, year, rent_price)
days = int(input("Enter the number of days you want to rent the vehicle: "))
total = vehicle1.calculate_rent(days)
print(f"The total rent price for the vehicle is: {total}")

car1 = Car("Toyota", "Corolla", 2022, 1500)
bike1 = Bike("Honda", "CD70", 2021, 500)

print(f"Car rent for 3 days: {car1.calculate_rent(3)}")
print(f"Bike rent for 5 days: {bike1.calculate_rent(5)}")
```

Vehicle Rental System

This program models a rental system for vehicles.

A Vehicle class holds details like make, model, year, and rent price. It has a method calculate_rent(days) to find the total cost.

Car and Bike classes inherit from Vehicle. They don’t add new features but prove the concept of inheritance.

Polymorphism is shown when we call calculate_rent() on both Car and Bike objects.
