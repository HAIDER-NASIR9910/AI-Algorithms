```python
from queue import PriorityQueue
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [], 'E': ['G'], 'F': ['G'], 'G': []
 }
 # Heuristic values (smaller = closer to goal)
heuristic = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 1,
    'E': 3,
    'F': 1,
    'G': 0
 }
 # Run the algorithm
def astar(start,goal):
  qeue=PriorityQueue()
  qeue.put((0,hurestic[start],0,start))
  while not qeue.empty():
    f,g,path=qeue.get()
    node=path[-1]
    if node==goal:
      return path
    for neighbour,cost in graph[node].items():
      new_g=g+cost
      new_f=new_g+heuristic[neighbour]
      qeue.put((new_f,new_g,path+[neighbour]))
  return None

astar("A","E")



```

```python
def best_first_search(graph,start,goal,hurestic):
  visited=set()
  queue=[start]
  while queue:
    queue.sort(key=lambda node:hurestic[node])
    node=queue.pop()
    if node==goal:
      print("visiting",node)
      return node
    for neighbour in graph[node]:
      if node  not in visited and neighbour not in queue:
        visited.add(node)
        queue.append(neighbour)

  return None
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [], 'E': ['G'], 'F': ['G'], 'G': []
 }
 # Heuristic values (smaller = closer to goal)
hurestic = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 1,
    'E': 3,
    'F': 1,
    'G': 0
 }
 # Run the algorithm
best_first_search(graph, 'A', 'G', hurestic)


```
