```python
#Haider Nasir
#B24S0308AI088
#AI-BLUE
#LAB#-8
from queue import PriorityQueue

def a_star_search(flights, heuristic, start, goal):
    open_list = PriorityQueue()
    open_list.put((0 + heuristic[start], 0, [start]))  # (f(n), g(n), path)
    visited = set()

    while not open_list.empty():
        f, g, path = open_list.get()
        current = path[-1]

        # Goal reached
        if current == goal:
            return path, g

        if current in visited:
            continue
        visited.add(current)

        for neighbor, cost in flights[current].items():
            new_g = g + cost
            f_new = new_g + heuristic[neighbor]
            open_list.put((f_new, new_g, path + [neighbor]))

    return None, float('inf')

# Example data
flights = {
    'A': {'B': 100, 'C': 300},
    'B': {'C': 100, 'D': 200},
    'C': {'D': 100, 'E': 400},
    'D': {'E': 100},
    'E': {}
}

heuristic = {'A': 400, 'B': 300, 'C': 150, 'D': 100, 'E': 0}

start, goal = 'A', 'E'

path, cost = a_star_search(flights, heuristic, start, goal)

print("Cheapest route:", path)
print("Total cost:", cost)
```

This program finds the cheapest route between airports using the A* algorithm.
It combines the real flight cost and the estimated remaining distance to pick the best path.
The algorithm explores routes step by step, always choosing the lowest total cost (f = g + h).
In the end, it gives the optimal route and total travel cost between two airports.

```python
from queue import PriorityQueue

def heuristic(a, b):
    # Manhattan Distance
    return abs(a[0] - b[0]) + abs(a[1] - b[1])

def a_star_grid(grid, start, goal):
    rows, cols = len(grid), len(grid[0])
    open_list = PriorityQueue()
    open_list.put((0 + heuristic(start, goal), 0, [start]))  # (f, g, path)
    visited = set()

    while not open_list.empty():
        f, g, path = open_list.get()
        current = path[-1]

        if current == goal:
            return path, g

        if current in visited:
            continue
        visited.add(current)

        x, y = current
        for dx, dy in [(0,1),(1,0),(-1,0),(0,-1)]:  # Right, Down, Up, Left
            nx, ny = x + dx, y + dy

            if 0 <= nx < rows and 0 <= ny < cols and grid[nx][ny] != '#':
                new_g = g + 1
                f_new = new_g + heuristic((nx, ny), goal)
                open_list.put((f_new, new_g, path + [(nx, ny)]))

    return None, float('inf')

# Example grid
grid = [
    ['S', '.', '.', '#', '.', '.'],
    ['.', '#', '.', '.', '.', '.'],
    ['.', '#', '.', '#', '.', '.'],
    ['.', '.', '.', '#', '.', 'G']
]

start = (0, 0)
goal = (3, 5)

path, cost = a_star_grid(grid, start, goal)

print("Shortest path:", path)
print("Steps required:", cost)
```

This program helps find the shortest path in a city grid, avoiding blocked roads.
Each road is represented by a grid cell, and movement is allowed in four directions.
The algorithm uses Manhattan distance to estimate how close it is to the destination.
It finds the quickest route from start (S) to goal (G) while skipping obstacles.
