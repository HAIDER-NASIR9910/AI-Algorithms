```python
friends = {

 "Alice": ["Bob", "Charlie", "Hannah"],
 "Bob": ["Alice", "David", "Eva", "Ian"],
 "Charlie": ["Alice", "Frank", "Grace"],
 "David": ["Bob", "Jack", "Karen"],
 "Eva": ["Bob", "Grace", "Liam"],
 "Frank": ["Charlie", "Mona"],
 "Grace": ["Charlie", "Eva", "Noah"],
 "Hannah": ["Alice", "Olivia"],
 "Ian": ["Bob", "Paul"],
 "Jack": ["David", "Quinn"],
 "Karen": ["David", "Rachel"],
 "Liam": ["Eva", "Sophia"],
 "Mona": ["Frank", "Tom"],
 "Noah": ["Grace", "Uma"],
 "Olivia": ["Hannah", "Victor"],
 "Paul": ["Ian", "Wendy"],
 "Quinn": ["Jack", "Xander"],
 "Rachel": ["Karen", "Yara"],
 "Sophia": ["Liam", "Zane"],
 "Tom": ["Mona"],
 "Uma": ["Noah"],
 "Victor": ["Olivia"],
 "Wendy": ["Paul"],
 "Xander": ["Quinn"],
 "Yara": ["Rachel"],
 "Zane": ["Sophia"]
}

visited = set()
def dfs(node):
    if node in visited:
        return
    print(node, " -> ")
    visited.add(node)
    for n in friends[node]:
        if n not in visited:
            dfs(n)

dfs("Alice")

```

This code is like exploring a friendship network starting from Alice.
We keep a notebook (`visited`) so we don’t meet the same friend twice.
At each step, we say hello (print the name) and then follow one of their friends deeper.
It keeps going until Alice’s entire circle of friends has been met.


```python
def iterative_dfs(start):
    visited = set()
    stack = [start]

    while stack:
        person = stack.pop()

        if person not in visited:
            print(person)
            visited.add(person)
            stack.extend(reversed(friends[person]))

    return visited

print("\nIterative DFS Traversal starting from Alice:")
iterative_dfs("Alice")
```

This code explores Alice’s network step by step using a stack.
Instead of diving in automatically (like recursion), we manually keep a “to-visit” list (stack).
Each time, we pull the last person out, say their name, and mark them visited.
Then we push their friends on the stack (reversed so the order looks like recursion), and continue until everyone is met.



```python
def dfs_find(start, target):
    visited = set()

    def helper(person):
        if person == target:
            print(f"Found {target} in the network")
            return True
        visited.add(person)
        for friend in friends[person]:
            if friend not in visited:
                if helper(friend):
                    return True
        return False

    if not helper(start):
        print("Friend not found")

f_name = input("Enter the person name you want to find: ")
print(f_name)
dfs_find("Alice", f_name)
```

It starts from Alice and checks if the person we’re searching for matches the target.
If not, it goes deeper through Alice’s friends one by one, keeping track of who’s already visited.
The moment the target is found, it prints a success message and stops exploring further.
If the whole network is explored without a match, it tells us the friend isn’t in the network.

```python
def dfs_count(start):
    visited = set()

    def helper(person):
        visited.add(person)
        for friend in friends[person]:
            if friend not in visited:
                helper(friend)

    helper(start)
    return len(visited)

print("\nNetwork size from Alice:", dfs_count("Alice"))

```

It begins with Alice and keeps a record (visited) of everyone already met.
Each time Alice meets a new friend, it marks them visited and goes deeper into their circle.
The process continues until no unvisited friends are left.
Finally, it returns the total number of unique people connected to Alice.

```python
def friend_circles():
    visited = set()
    circles = []

    for person in friends:
        if person not in visited:
            circle = []
            def helper(p):
                visited.add(p)
                circle.append(p)
                for f in friends[p]:
                    if f not in visited:
                        helper(f)
            helper(person)
            circles.append(circle)

    return circles

print("\nFriend Circles in the network:")
for i, circle in enumerate(friend_circles(), 1):
    print(f"Circle {i}: {circle}")
```

It goes through every person in the network and checks if they’ve already been visited.
If not, it starts exploring their whole circle of friends using DFS and groups them together.
Each fully explored group is saved as one “friend circle.”
At the end, it prints all the circles — showing who belongs together in the same connected group.
