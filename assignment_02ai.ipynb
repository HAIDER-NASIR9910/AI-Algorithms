This code performs a **Depth-First Search (DFS)** to find a path from a starting node to a goal node in a **weighted graph**.
It uses a **stack** to explore one path deeply before moving to the next, keeping track of **visited nodes** and **nodes generated**.
In simple words — it’s like exploring a maze by going as deep as possible down one route before backtracking to try another, until it finds the target node.


```python
TASK:4(3)
import heapq
def uniform_cost_search(graph,start,goal):
  visited=set()
  pq=[(0,[start])]
  nodes_generated = 1
  while pq:
    cost,path=heapq.heappop(pq)
    node=path[-1]
    if node==goal:
      print("Total number of nodes_generated",nodes_generated,"Queue size is",len(pq))
      return cost,path,nodes_generated,len(pq)

    if node not in visited:
      visited.add(node)
      for neighbour,cost_edge in graph.get(node,{}).items():
        new_cost=cost+cost_edge
        new_path=path+[neighbour]
        heapq.heappush(pq,(new_cost,new_path))
        nodes_generated += 1
  return None

graph = {
    'A': {'B': 3, 'C': 4},
    'B': {'A': 3, 'E': 10, 'I': 11, 'H': 10},
    'C': {'A': 4, 'D': 7, 'E': 7},
    'D': {'C': 7, 'E': 9, 'F': 9},
    'E': {'B': 10, 'C': 7, 'D': 9, 'F': 11, 'H': 13},
    'F': {'D': 9, 'E': 11, 'G': 13},
    'G': {'F': 13, 'H': 18, 'K': 18},
    'H': {'E': 13, 'B': 10, 'I': 10, 'J': 18, 'K': 18, 'L': 20},
    'I': {'B': 11, 'H': 10, 'J': 19},
    'J': {'I': 19, 'H': 18},
    'K': {'G': 18, 'H': 18, 'L': 19},
    'L': {'H': 20, 'K': 19}
}
result = uniform_cost_search(graph, 'A', 'E')
print(result)
```

This program finds the cheapest path between two points in a weighted graph using the Uniform Cost Search (UCS) algorithm.

It starts from a chosen node (here 'A') and explores paths step by step, always picking the one with the lowest total cost so far.

The heapq library is used to manage the priority queue — meaning nodes with smaller travel costs are visited first.

As the algorithm runs, it keeps track of how many nodes were generated and how many are still waiting in the queue.

Once it reaches the goal node ('E'), it prints the total cost, the path taken, and some search statistics.

In simple terms — it’s like finding the least expensive route from point A to point E on a map, checking every option carefully but smartly choosing the cheaper roads first.

```python
TASK:3
class Faisalabad_karachi:
  def __init__(self,Distance):
    self.Distance=Distance
  def fly(self):
    print("If you want to fly total time will be:")
    return self.Distance/800
  def road(self):
    print("IF YOU WANT TO TRAVEL BY ROAD TIME WILL BE")
    return self.Distance/100



trip = Faisalabad_karachi(1115.3)  # distance in km
choice=input("If you want to travel by road or fly?")
if choice=="road":
  print(trip.fly(), "hours")
else:
  print(trip.road(), "hours")
```
