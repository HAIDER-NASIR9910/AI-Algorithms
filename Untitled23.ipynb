```python
# upgraded_drone_ccp.py
import math
import time
from collections import deque, defaultdict
import heapq
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# ===================== GRID AND COSTS =====================
grid = [
    ['B','P','P','P','F','F','F','X','X','P','P','P','F','F','H','P','F','P','P','P'],
    ['P','P','F','F','H','F','F','X','X','P','F','H','H','P','P','F','P','F','F','P'],
    ['P','F','F','H','F','F','P','X','X','P','H','H','F','F','H','P','F','P','P','P'],
    ['P','F','H','H','F','P','P','X','X','X','P','F','H','H','P','P','P','F','P','P'],
    ['F','H','H','F','F','P','R','X','X','X','P','F','F','H','H','P','F','P','P','F'],
    ['H','H','F','F','P','P','P','P','P','P','P','F','H','F','F','H','P','P','P','P'],
    ['P','P','F','H','H','F','F','F','P','V','P','P','P','H','F','P','P','F','P','P'],
    ['F','F','H','F','F','F','H','H','F','P','P','H','P','F','P','P','F','P','H','P'],
    ['P','F','H','H','F','P','P','P','P','P','P','P','F','H','F','F','P','P','P','P'],
    ['P','P','F','F','P','R','H','P','P','P','P','P','P','H','H','F','F','P','F','P'],
    ['P','H','H','F','F','F','F','F','P','P','P','H','F','F','P','P','H','P','P','P'],
    ['P','F','F','P','P','P','H','H','F','F','P','P','P','F','F','P','F','F','H','P'],
    ['P','P','H','F','F','P','P','P','H','H','P','P','P','P','F','H','P','P','F','P'],
    ['P','F','H','H','P','P','F','H','F','P','P','H','P','F','F','H','P','V','P','P'],
    ['P','P','F','P','P','P','H','F','H','P','F','P','P','P','H','F','P','P','P','P'],
    ['P','P','P','P','F','P','F','P','H','P','P','P','P','F','P','H','P','F','F','P'],
    ['P','F','H','P','P','P','F','F','P','P','P','P','F','H','P','P','V','P','P','P'],
    ['P','F','F','H','P','P','H','F','F','P','H','P','P','P','P','P','F','H','P','P'],
    ['P','P','P','P','P','H','P','P','P','H','F','H','P','V','P','R','P','P','P','P'],
    ['P','P','P','F','P','P','P','F','F','P','P','P','H','F','H','P','P','P','V','P']
]

costs = {'P':1, 'F':2, 'H':3, 'B':0, 'V':1, 'R':1, 'X':math.inf}
# Fuel limit chosen for this script: 25 (change if you want 20)
fuel_limit = 25

rows, cols = len(grid), len(grid[0])

# ===================== HELPER FUNCTIONS =====================
def get_neighbors(r, c):
    moves = [(1,0), (-1,0), (0,1), (0,-1)]
    for dr, dc in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] != 'X':
            yield nr, nc

def terrain_cost(r, c):
    return costs.get(grid[r][c], math.inf)

def manhattan(a, b):
    return abs(a[0]-b[0]) + abs(a[1]-b[1])

def euclidean(a, b):
    return math.sqrt((a[0]-b[0])**2 + (a[1]-b[1])**2)

# find coordinates of special nodes: base, villages, refuels
def find_nodes():
    base = None
    villages = []
    refuels = []
    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == 'B':
                base = (r,c)
            elif grid[r][c] == 'V':
                villages.append((r,c))
            elif grid[r][c] == 'R':
                refuels.append((r,c))
    return base, villages, refuels

base, villages, refuels = find_nodes()
# If grid uses 'V' not named indices, they are villages in order encountered; match original problem mapping roughly.
# For user clarity, print found nodes:
print("Base:", base)
print("Villages:", villages)
print("Refuels:", refuels)
print("Fuel limit per drone:", fuel_limit)

def astar(start, goal, method="manhattan", fuel_limit_override=None):

    if start == goal:
        return [start], 0, 0

    def h(a, b):
        if method == "manhattan":
            return manhattan(a, b) * 1
        elif method == "euclidean":
            return euclidean(a,b)
        elif method == "custom":

            return manhattan(a,b) * 1
        else:
            return 0

    open_heap = []
    heapq.heappush(open_heap, (h(start, goal), start))
    came_from = {}
    gscore = {start: 0}
    closed = set()
    nodes_expanded = 0

    while open_heap:
        f, current = heapq.heappop(open_heap)
        nodes_expanded += 1
        if current == goal:
            # reconstruct path
            path = []
            node = current
            while node in came_from:
                path.append(node)
                node = came_from[node]
            path.append(start)
            path.reverse()

            path_cost = sum(terrain_cost(r,c) for (r,c) in path[1:])
            return path, path_cost, nodes_expanded

        if current in closed:
            continue
        closed.add(current)

        r, c = current
        for nr, nc in get_neighbors(r,c):
            neighbor = (nr, nc)
            tentative_g = gscore[current] + terrain_cost(nr, nc)
            if fuel_limit_override is not None and tentative_g > fuel_limit_override:
                continue
            if neighbor not in gscore or tentative_g < gscore[neighbor]:
                gscore[neighbor] = tentative_g
                came_from[neighbor] = current
                heapq.heappush(open_heap, (tentative_g + h(neighbor, goal), neighbor))

    return None, math.inf, nodes_expanded

def dijkstra(start, goal):
    if start == goal:
        return [start], 0, 0
    pq = []
    heapq.heappush(pq, (0, start))
    dist = {start: 0}
    came_from = {}
    visited = set()
    nodes_expanded = 0
    while pq:
        d, node = heapq.heappop(pq)
        nodes_expanded += 1
        if node == goal:
            # reconstruct path
            path = []
            cur = node
            while cur in came_from:
                path.append(cur)
                cur = came_from[cur]
            path.append(start)
            path.reverse()
            return path, d, nodes_expanded
        if node in visited:
            continue
        visited.add(node)
        r, c = node
        for nr, nc in get_neighbors(r,c):
            nd = d + terrain_cost(nr, nc)
            if (nr, nc) not in dist or nd < dist[(nr, nc)]:
                dist[(nr, nc)] = nd
                came_from[(nr, nc)] = node
                heapq.heappush(pq, (nd, (nr, nc)))
    return None, math.inf, nodes_expanded
def bfs(start, goal):
    q = deque([(start, [start])])
    visited = set([start])
    nodes_expanded = 0
    while q:
        (r, c), path = q.popleft()
        nodes_expanded += 1
        if (r, c) == goal:
            path_cost = sum(terrain_cost(r,c) for r,c in path[1:])
            return path, path_cost, nodes_expanded, visited
        for nr, nc in get_neighbors(r, c):
            if (nr, nc) not in visited:
                visited.add((nr, nc))
                q.append(((nr, nc), path + [(nr, nc)]))
    return None, math.inf, nodes_expanded, visited

def dfs(start, goal):
    stack = [(start, [start])]
    visited = set([start])
    nodes_expanded = 0
    while stack:
        (r, c), path = stack.pop()
        nodes_expanded += 1
        if (r, c) == goal:
            path_cost = sum(terrain_cost(r,c) for r,c in path[1:])
            return path, path_cost, nodes_expanded, visited
        for nr, nc in get_neighbors(r, c):
            if (nr, nc) not in visited:
                visited.add((nr, nc))
                stack.append(((nr, nc), path + [(nr, nc)]))
    return None, math.inf, nodes_expanded, visited
pairwise_cache = {}

def shortest_cost_and_path(a,b):
    # cache symmetric pairs
    key = (a,b)
    if key in pairwise_cache:
        return pairwise_cache[key]
    path, cost, nodes = dijkstra(a,b)
    pairwise_cache[key] = (path,cost,nodes)
    return path,cost,nodes

# Greedy assignment: sort villages by cost from base then assign to the currently less-loaded drone
def greedy_assign(base, villages):
    # Compute base->v cost
    entries = []
    for v in villages:
        _, cost, _ = shortest_cost_and_path(base, v)
        entries.append((v, cost))
    # sort nearest-first
    entries.sort(key=lambda x: x[1])
    drones = {0: {'assigned': [], 'load': 0}, 1: {'assigned': [], 'load': 0}}
    for v, c in entries:
        # assign to drone with smaller load
        if drones[0]['load'] <= drones[1]['load']:
            drones[0]['assigned'].append(v)
            drones[0]['load'] += c
        else:
            drones[1]['assigned'].append(v)
            drones[1]['load'] += c
    return drones

# attempt to plan a leg from 'start' to 'goal' with remaining_fuel.
# Returns list of path segments (each is list of coords), total_cost, nodes_expanded_total, success_flag, refuels_used
def plan_leg_with_refuel(start, goal, remaining_fuel):
    # First try direct A* with fuel constraint
    path, cost, nodes = astar(start, goal, method="manhattan", fuel_limit_override=remaining_fuel)
    if path is not None:
        return [path], cost, nodes, True, []
    # otherwise try to insert one refuel stop (nearest R) that is reachable and allows completion
    best_solution = None
    for rnode in refuels:
        # start -> R must be reachable with remaining_fuel
        p1, c1, n1 = astar(start, rnode, method="manhattan", fuel_limit_override=remaining_fuel)
        if p1 is None:
            continue
        # after refueling, we have full fuel_limit for R -> goal
        p2, c2, n2 = astar(rnode, goal, method="manhattan", fuel_limit_override=fuel_limit)
        if p2 is None:
            continue
        total_cost = c1 + c2
        total_nodes = n1 + n2
        # Accept first feasible; could pick best by nodes or cost; choose minimal total_cost
        if best_solution is None or total_cost < best_solution[1]:
            best_solution = ([p1, p2], total_cost, total_nodes, [rnode])
    if best_solution:
        return best_solution[0], best_solution[1], best_solution[2], True, best_solution[3]
    # if not possible with single refuel, we could attempt multiple refuels (omitted for brevity) -- mark as fail
    return [], math.inf, 0, False, []

# Plan full route for assigned list: start from base, visit assigned villages in order, return to base.
# Greedy visiting order: simply in assigned order; could be improved with TSP heuristics.
def plan_drone_mission(drone_id, assigned_villages):
    report = {
        'drone': drone_id,
        'legs': [],  # each leg: dict with start, goal, path_segments, cost, nodes_expanded, success, refuels
        'total_cost': 0,
        'total_nodes_expanded': 0,
        'steps': 0,
        'success': True,
        'execution_time': 0.0
    }
    start_time = time.perf_counter()
    cur_pos = base
    remaining_fuel = fuel_limit
    total_cost = 0
    total_nodes = 0
    all_path_cells = []  # for heatmap marking
    for v in assigned_villages:
        segments, cost, nodes_expanded, success, refuels_used = plan_leg_with_refuel(cur_pos, v, remaining_fuel)
        leg = {
            'start': cur_pos,
            'goal': v,
            'segments': segments,
            'cost': cost,
            'nodes_expanded': nodes_expanded,
            'success': success,
            'refuels': refuels_used
        }
        report['legs'].append(leg)
        if not success:
            report['success'] = False
            # mark this village unserved and continue (or break) -- here we continue to attempt others
            cur_pos = cur_pos  # stays same
            continue
        # apply fuel usage: subtract cost of first segment from remaining_fuel; then if refueled mid-leg, reset
        if len(segments) == 1:
            # direct leg
            remaining_fuel -= cost
        else:
            # segments includes a refuel (we modeled start->R then R->goal)
            # For the first segment (start->R), subtract its cost then refuel -> remaining_fuel becomes fuel_limit - cost(second segment)
            c1 = sum(terrain_cost(r,c) for r,c in segments[0][1:]) if segments[0] else 0
            c2 = sum(terrain_cost(r,c) for r,c in segments[1][1:]) if segments[1] else 0
            remaining_fuel -= c1
            # refuel
            remaining_fuel = fuel_limit - c2
        total_cost += cost
        total_nodes += nodes_expanded
        for seg in segments:
            for cell in seg:
                all_path_cells.append(cell)
        # move drone to village location (after successful leg)
        cur_pos = v
    # After visiting assigned villages, plan return to base
    segments, cost, nodes_expanded, success, refuels_used = plan_leg_with_refuel(cur_pos, base, remaining_fuel)
    leg = {
        'start': cur_pos,
        'goal': base,
        'segments': segments,
        'cost': cost,
        'nodes_expanded': nodes_expanded,
        'success': success,
        'refuels': refuels_used
    }
    report['legs'].append(leg)
    if not success:
        report['success'] = False
    else:
        # update totals similarly
        if len(segments) == 1:
            remaining_fuel -= cost
        else:
            c1 = sum(terrain_cost(r,c) for r,c in segments[0][1:]) if segments[0] else 0
            c2 = sum(terrain_cost(r,c) for r,c in segments[1][1:]) if segments[1] else 0
            remaining_fuel -= c1
            remaining_fuel = fuel_limit - c2
        total_cost += cost
        total_nodes += nodes_expanded
        for seg in segments:
            for cell in seg:
                all_path_cells.append(cell)

    report['total_cost'] = total_cost
    report['total_nodes_expanded'] = total_nodes
    report['steps'] = len(set(all_path_cells))
    report['execution_time'] = time.perf_counter() - start_time
    return report, all_path_cells

# ===================== RUN THE MULTI-DRONE MISSION =====================

# Assignment
drones_assignment = greedy_assign(base, villages)
print("\nGreedy assignment result:")
for d in drones_assignment:
    print(f"Drone {d} assigned villages: {drones_assignment[d]['assigned']} (estimated load {drones_assignment[d]['load']:.2f})")

# Plan missions for both drones
drone_reports = {}
heatmap_total = np.zeros((rows, cols))
for d in [0,1]:
    assigned = drones_assignment[d]['assigned']
    report, path_cells = plan_drone_mission(d, assigned)
    drone_reports[d] = report
    # mark visited/path cells in heatmap_total for visualization
    for (r,c) in path_cells:
        heatmap_total[r][c] = min(1.0, heatmap_total[r][c] + 0.8)  # stack influence
    # print short summary
    print(f"\nDrone {d} report:")
    print(f"  Assigned: {assigned}")
    print(f"  Success: {report['success']}")
    print(f"  Total cost: {report['total_cost']:.2f}")
    print(f"  Nodes expanded: {report['total_nodes_expanded']}")
    print(f"  Steps (unique path cells): {report['steps']}")
    print(f"  Execution time: {report['execution_time']:.4f}s")
    # detail legs
    for i, leg in enumerate(report['legs']):
        print(f"    Leg {i+1}: {leg['start']} -> {leg['goal']}, success={leg['success']}, cost={leg['cost']}, nodes={leg['nodes_expanded']}, refuels={leg['refuels']}")

# ===================== RESULTS TABLE =====================
rows_out = []
for d, rep in drone_reports.items():
    rows_out.append({
        'Drone': d,
        'Assigned': ','.join([str(x) for x in drones_assignment[d]['assigned']]),
        'Success': rep['success'],
        'TotalCost': rep['total_cost'],
        'NodesExpanded': rep['total_nodes_expanded'],
        'UniqueSteps': rep['steps'],
        'Time(s)': round(rep['execution_time'], 4)
    })

df_results = pd.DataFrame(rows_out)
print("\nMission Summary Table:")
print(df_results)

# ===================== VISUALIZATIONS =====================

# Heatmap of combined paths
plt.figure(figsize=(6,6))
plt.title("Combined Drone Paths Heatmap")
plt.imshow(heatmap_total, cmap='YlOrRd', origin='upper')
plt.colorbar(label='Exploration / Path intensity')
plt.scatter([base[1]],[base[0]], marker='*', color='black', s=100, label='Base')
vx = [v[1] for v in villages]
vy = [v[0] for v in villages]
plt.scatter(vx, vy, marker='s', color='blue', s=60, label='Villages')
rx = [r[1] for r in refuels]
ry = [r[0] for r in refuels]
plt.scatter(rx, ry, marker='o', facecolors='none', edgecolors='green', s=80, label='Refuels')
plt.legend(loc='upper right', bbox_to_anchor=(1.35, 1.0))
plt.show()

# Bar plots
plt.figure(figsize=(8,4))
plt.bar(df_results['Drone'].astype(str), df_results['TotalCost'])
plt.xlabel('Drone')
plt.ylabel('Total Cost')
plt.title('Per-Drone Total Cost')
plt.show()

plt.figure(figsize=(8,4))
plt.bar(df_results['Drone'].astype(str), df_results['NodesExpanded'])
plt.xlabel('Drone')
plt.ylabel('Nodes Expanded')
plt.title('Per-Drone Nodes Expanded')
plt.show()

# Per-leg visualization: small heatmap per drone
for d, rep in drone_reports.items():
    hm = np.zeros((rows, cols))
    for leg in rep['legs']:
        for seg in leg['segments']:
            for (r,c) in seg:
                hm[r][c] = 1.0
    plt.figure(figsize=(5,5))
    plt.title(f"Drone {d} Path Cells")
    plt.imshow(hm, cmap='YlGnBu')
    plt.scatter([base[1]],[base[0]], marker='*', color='black', s=80)
    plt.show()

print("\nEnd of upgraded multi-drone mission planning run.")
```
